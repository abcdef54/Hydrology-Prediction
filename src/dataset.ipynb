````python
import os
import pandas as pd
from pathlib import Path
````

````python
DIR_REPORTS = "../reports/"

DATA_PATH_10MIN = os.path.join(DIR_REPORTS, "hydrology_training_10min_v2.csv")
DATA_PATH_30MIN = os.path.join(DIR_REPORTS, "hydrology_training_30min.csv")
DATA_PATH_1H = os.path.join(DIR_REPORTS, "hydrology_training_1h.csv")


print(os.path.exists(DATA_PATH_10MIN))
print(os.path.exists(DATA_PATH_30MIN))
print(os.path.exists(DATA_PATH_1H))

SAVE_PATH = "./train_data/"

if not os.path.exists(SAVE_PATH):
    os.makedirs(SAVE_PATH)
````

## ds 10min

````python
targets = [
    "target_water_level_plus_60m",
    "target_water_level_plus_180m",
    "target_water_level_plus_360m",
]

output_dir = Path("../train_data/10min")
output_dir.mkdir(parents=True, exist_ok=True)

ds_30min = pd.read_csv(
    DATA_PATH_10MIN,
    parse_dates=["timestamp_utc"],
)

model_df = (
    ds_30min
    .sort_values("timestamp_utc")
    .drop_duplicates("timestamp_utc")
    .dropna(subset=targets)
    .reset_index(drop=True)
)

train_ratio = 0.70
val_ratio = 0.15
test_ratio = 0.15

assert abs(train_ratio + val_ratio + test_ratio - 1.0) < 1e-9

total = len(model_df)
train_end = int(total * train_ratio)
val_end = int(total * (train_ratio + val_ratio))

validation_start = model_df.iloc[train_end]["timestamp_utc"]
test_start = model_df.iloc[val_end]["timestamp_utc"]

# Largest target horizon.
max_horizon = pd.Timedelta(hours=24)

train_candidates = model_df.iloc[:train_end]
validation_candidates = model_df.iloc[train_end:val_end]
test_split = model_df.iloc[val_end:].copy()

# Ensure every training target occurs before validation starts.
train_split = train_candidates.loc[
    train_candidates["timestamp_utc"] + max_horizon < validation_start
].copy()

# Ensure every validation target occurs before test starts.
val_split = validation_candidates.loc[
    validation_candidates["timestamp_utc"] + max_horizon < test_start
].copy()

train_split.to_csv(output_dir / "train.csv", index=False)
val_split.to_csv(output_dir / "val.csv", index=False)
test_split.to_csv(output_dir / "test.csv", index=False)

print("Train:", len(train_split))
print("Validation:", len(val_split))
print("Test:", len(test_split))
````

### 1. How Many Parameters Are You Training?

When you call `freeze_backbone()`, **only the three adapter modules are trained** (`hindcast_adapter`, `forecast_adapter`, and `output_head`). All backbone parameters have `requires_grad = False`.

#### Exact Parameter Formulas

| Component | Layer | Weight Parameters | Bias / Affine Parameters | Total |
| :--- | :--- | :--- | :--- | :--- |
| **`hindcast_adapter`** | `nn.Linear` | $N_{\text{hind\_local}} \times N_{\text{hind\_pre}}$ | $N_{\text{hind\_pre}}$ | $(N_{\text{hind\_local}} + 1) \times N_{\text{hind\_pre}}$ |
| | `nn.LayerNorm` | — | $2 \times N_{\text{hind\_pre}}$ (gain $\gamma$ + bias $\beta$) | $2 \times N_{\text{hind\_pre}}$ |
| **Subtotal `hindcast_adapter`** | | | | **$(N_{\text{hind\_local}} + 3) \times N_{\text{hind\_pre}}$** |
| **`forecast_adapter`** | `nn.Linear` | $N_{\text{fore\_local}} \times N_{\text{fore\_pre}}$ | $N_{\text{fore\_pre}}$ | $(N_{\text{fore\_local}} + 1) \times N_{\text{fore\_pre}}$ |
| | `nn.LayerNorm` | — | $2 \times N_{\text{fore\_pre}}$ (gain $\gamma$ + bias $\beta$) | $2 \times N_{\text{fore\_pre}}$ |
| **Subtotal `forecast_adapter`** | | | | **$(N_{\text{fore\_local}} + 3) \times N_{\text{fore\_pre}}$** |
| **`output_head`** | `nn.Linear` | $H \times N_{\text{out}}$ | $N_{\text{out}}$ | **$(H + 1) \times N_{\text{out}}$** |

$$\text{Total Trainable Params} = (N_{\text{hind\_local}} + 3) \times N_{\text{hind\_pre}} + (N_{\text{fore\_local}} + 3) \times N_{\text{fore\_pre}} + (H + 1) \times N_{\text{out}}$$

Where:
- $N_{\text{hind\_local}}, N_{\text{fore\_local}}$: local feature dimensions (in your dataset: up to **178** features)
- $N_{\text{hind\_pre}}, N_{\text{fore\_pre}}$: backbone input dimensions (typically **32** or **64** in OpenHydroNet)
- $H$: backbone `hidden_size` (typically **64**, **128**, or **256**)
- $N_{\text{out}}$: downstream target count (**3** for `target_water_level_plus_60m`, `180m`, `360m`)

---

#### Concrete Example for Your Dataset

Assuming:
- Local features: $N_{\text{hind\_local}} = 178$, $N_{\text{fore\_local}} = 178$
- Pretrained input dimensions: $N_{\text{hind\_pre}} = 64$, $N_{\text{fore\_pre}} = 64$
- Pretrained hidden size: $H = 128$
- Targets: $N_{\text{out}} = 3$

1. **`hindcast_adapter`**: $(178 + 3) \times 64 = \mathbf{11,584}$
2. **`forecast_adapter`**: $(178 + 3) \times 64 = \mathbf{11,584}$
3. **`output_head`**: $(128 + 1) \times 3 = \mathbf{387}$
4. **Total Trainable**: $\mathbf{23,555}$ parameters (typically $< 5\%$ to $10\%$ of the full model).

You can verify the exact parameter count programmatically at runtime with:

```python
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
total_params = sum(p.numel() for p in model.parameters())
print(f"Trainable: {trainable_params:,} / Total: {total_params:,} ({trainable_params/total_params:.2%})")
```

### 2. Is This Similar to LoRA?

**No, this is fundamentally different from LoRA.**

Here is the key architectural difference:

```
=== LoRA (Low-Rank Adaptation) ===
Input x  (must ALREADY match pretrained dimension d)
   │
   ├───> [ Frozen W (d x k) ] ────┐
   │                              (+) ───> Output y (k)
   └───> [ A (d x r) ] ──> [ B (r x k) ] ─┘   (r << min(d, k))


=== Your Architecture (Feature Alignment Projector + Head) ===
Local Inputs (178) ───> [ Hindcast Adapter ] ───> (64) Pretrained Space
                                                       │
                                            [ Frozen OpenHydroNet Backbone ]
                                                       │
Local Targets (3)  <─── [ Output Head ]     <─── (128) Hidden State
```

| Dimension | LoRA (Low-Rank Adaptation) | Your Current Setup |
| :--- | :--- | :--- |
| **Where it sits** | **Inside** internal layers ($W_{ih}$, $W_{hh}$, attention matrices). | **At the outer boundaries** (input and output interfaces). |
| **Mathematical operation** | $W' = W_{\text{frozen}} + \frac{\alpha}{r}(B \times A)$. Modifies internal linear mappings via low-rank residuals. | Affine projection: $y = \text{LayerNorm}(W x + b)$. |
| **Input compatibility** | Assumes the input tokens/features already have identical meaning and dimensions. | Solves **domain/feature mismatch**: converts 178 local telemetry signals into OpenHydroNet dimensions. |
| **Backbone behavior** | The backbone's internal representations are altered by the rank updates. | The backbone is a completely frozen black-box feature extractor. |
| **Analogy in Deep Learning** | LoRA in LLMs / Stable Diffusion. | **Multimodal Projectors** (e.g., the MLP projector in LLaVA that maps CLIP visual vectors into LLM token space) + **Linear Probing**. |

### Summary

What you are building here is an **Input Alignment Projector + Transfer Head** (often termed a *Feature Adapter* or *Domain Adapter*), rather than parameter-efficient fine-tuning (PEFT/LoRA). It allows you to leverage OpenHydroNet's temporal recurrent dynamics without needing OpenHydroNet's original ERA5/GFS meteorological inputs or discharge targets.

### 1. Relative to the River's Dynamic Range: 15 cm is Actually Very Strong

From the exploratory data analysis of Dã Viên station ([hydrology_training_10min_v2.csv](file:///home/my_ubuntu/projects/hydrology/reports/hydrology_training_10min_v2.csv)):
- **Mean water level**: $\sim 483\text{ cm}$
- **Standard deviation ($\sigma$)**: **$208.7\text{ cm}$**
- **Min $\to$ Max Range**: **$47.8\text{ cm} \to 3011.8\text{ cm}$** (a span of nearly $30\text{ meters}$)

An RMSE of **$15.5\text{ cm}$** against a standard deviation of $209\text{ cm}$ yields:
$$\text{NRMSE} = \frac{15.5}{208.7} \approx 7.4\% \quad (R^2 > 0.99)$$

For a complex tropical river system influenced by upstream dam releases, tidal backwater, and intense rainfall, an overall error band of $\pm 15\text{ cm}$ on the 1-hour horizon is a high-accuracy fit.

### 4. Where Your Concern *Is* Completely Justified

You are right to be cautious for three key reasons:

1. **Longer Lead Times Degradation**:
   While the error is **$15\text{ cm}$ at +1h**, it rises to **$51.6\text{ cm}$ at +6h** and **$64\text{ cm}$ at +24h**. 
   An error of **$64\text{ cm}$** is indeed large enough to cross warning levels (e.g. predicting Báo động I when the real stage reaches Báo động II).
2. **Global RMSE Hides Peak Errors (Peak Smoothing)**:
   The dataset consists of $>90\%$ dry/normal baseflow days and only a few extreme storm events. Because RMSE averages across all timesteps, tree models tend to predict the average well, but **underestimate the extreme crest of the flood peak by much more than $15\text{ cm}$**.
3. **High-Water Test Slices**:
   When evaluating baselines specifically on high-water intervals (`high_water_test` in [baseline_metrics.csv](file:///home/my_ubuntu/projects/hydrology/reports/baseline_metrics.csv)), error typically doubles or triples compared to the full test set.

> [!TIP]
> To verify whether the error during actual flood events is acceptable, evaluate peak-specific metrics (e.g., **Peak Absolute Error** and **Peak Timing Error**) on flood subsets, as implemented in [scripts/run_baselines.py](file:///home/my_ubuntu/projects/hydrology/scripts/run_baselines.py#L95-L100).
